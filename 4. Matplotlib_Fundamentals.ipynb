# **Matplotlib For ML**

```python
import matplotlib.pyplot as plt
accuracy = [0.61,0.72,0.81, 0.87,0.91]
plt.plot(accuracy)
plt.show()
```

```python
import matplotlib.pyplot as plt
x=[1,2,3,4,5]
y=[10,20,30,40,50]
plt.plot(x,y)
plt.show()
```

***Figure & Axes***

```python
fig,ax = plt.subplots()
ax.plot(
    [1,2,3,4,5],
    [10,20,30,40,50]
)

# Figure = Whole Canvas
# Axes = individual Plot
```

```python
plt.plot(x,y)
plt.title("Sales")
plt.xlabel("Month")
plt.ylabel("Revenue")
plt.show()
```

# **LINE** Plot

```python
from matplotlib.lines import lineStyles
from matplotlib import colors
epochs =[1,2,3,4,5]
loss =[0.01,0.68,0.51,0.39,0.31]
plt.plot(epochs,loss, marker='*', color='royalblue',linewidth=2)
plt.title("Training Loss")
plt.xlabel("Epoch")
plt.ylabel("Loss")
plt.grid(True, linestyle='--',alpha=0.3)
plt.show()
```

# **BAR Chart**

```python
models = [
    "Logistic",
    "Random Forest",
    "XGBoost"
]

accuracy = [
    0.82,0.89,0.92
]
plt.bar(models, accuracy)
plt.title("Model Accuracy Comparision")
plt.ylabel("Accuracy")
plt.show()
```

**Horizontal Bar Chart**

```python
features =[
    "Monthly_Charges",
    "Support_Tickets",
    "Tenure",
    "Payment_Failures",
    "age"
]

importance = [
    0.31,
    0.26,
    0.22,
    0.14,
    0.07
]
plt.barh(features,importance)
plt.title("Feature Importance")
plt.ylabel("Importance")
plt.show()
```

# **Scatter Plot**

```python
features =[
    "Monthly_Charges",
    "Support_Tickets",
    "Tenure",
    "Payment_Failures",
    "age"
]

importance = [
    0.31,
    0.26,
    0.22,
    0.14,
    0.07
]
plt.scatter(features,importance)
plt.title("Feature Importance")
plt.ylabel("Importance")
plt.show()
```

# **HISTOGRAM**

```python
ages =[
    22,23,24,25,
    27,28,29,30,
    31,31,32,35,
    38,42,45,48
]

plt.hist(ages,bins=6)
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.title("Customer Age Distribution")
plt.show()
```

# **Pie Chart**

```python
labels = [
    "churn",
    "No Churn"
]
sizes =[
    28,72
]
plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%"
)
plt.title("Customer Churn")
plt.show()
```

# `imshow()`

```python
import numpy as np
import matplotlib.pyplot as plt

image = np.random.random(
    (28,28,3)
)
plt.imshow(
    image
)
plt.title("Exampl Image")
plt.show()
```

# **When to Use Which Plot ?**

```python
# One Numeric = Histogram / Box Plot

# Two Numeric = Scatter Plot

# categories = BAR / Count Plot

# Time / Ordered Data = Line Plot

# Matrix /Image = Heatmap /imshow()
```

# **PANDAS with MATPLOTLIB**

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.DataFrame({
    "age":
    [22,25,28,31,
    35,41,45,50],
    "tenure":
    [
        1,3,5,8,
        12,24,36,60
    ]

})

plt.hist(
    df["age"],
    bins=5
)
plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Tenure")
plt.show()
```
