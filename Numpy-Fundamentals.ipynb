**SELECTING DATA FROM MATRIX**

```python
X = np.array([
    [25,50000,2],
    [32,75000,5],
    [45,90000,8]
])

print(X[0]) # First
print(X[0,1]) # 0th index row and 1st index value
print(X[:,0]) # all in the 0th index colmns
print(X[1,:])  # all from 1st Index
```

# **AGGREGATION**

```python
scores= np.array([0.81,0.87,0.92,0.89])

print(np.mean(scores)) # MEAN : Calculates the arithmetic average of all scores.
print(np.median(scores)) # MEDIAN : Finds the middle value when scores are sorted from lowest to highest.
print(np.min(scores)) # Min Value : Identifies the lowest score in the dataset.
print(np.max(scores)) # MAX values : Identifies the highest score in the dataset.
print(np.sum(scores)) # Total SUM : Adds all the scores together to get the grand total.
print(np.std(scores)) # Standard Deviations : Measures the average distance of the scores from the mean.
print(np.var(scores)) # Variance : Measures the spread of data by averaging the squared distances from the mean.
```

**Understanding** `axis`

```python
X= np.array([
    [10,20],
    [30,40],
    [50,60]
])

print(np.mean(X, axis=0))
# axis=0 means the operation runs downwards across rows (collapsing the rows to calculate a result for each column).
# axis=1 means the operation runs horizontally across columns (collapsing the columns to calculate a result for each row).
```

# **RESHAPING ARRAYS**

```python
a = np.arange(12)
print(a.shape)
matrix = a.reshape(3,4)
print(matrix)
print(matrix.flatten()) # Creates a brand new copy of the data in memory. If you modify the flattened array, the original matrix does not change
print(matrix.ravel()) # Creates a view of the original array whenever possible without copying data. If you modify the raveled array, the original matrix changes too.
```

```python
import numpy as np
X = np.array([[10, 20], [30, 40], [50, 60]])

# Both look identical when printed:
print(X.flatten()) # Output: [10 20 30 40 50 60]
print(X.ravel())   # Output: [10 20 30 40 50 60]

# Memory behavior difference:
flat_array = X.flatten()
flat_array[0] = 999  # X is STILL [[10, 20], [30, 40], [50, 60]]

rav_array = X.ravel()
rav_array[0] = 999   # X is NOW [[999, 20], [30, 40], [50, 60]]
```


1.  Use **flatten**() if you need a safe copy that won't mess up your original data.
2.   Use **ravel**() if you want maximum speed and memory efficiency, as it does not duplicate data.
