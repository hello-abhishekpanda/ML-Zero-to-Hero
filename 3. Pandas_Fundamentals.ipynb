**SERIES | DATA FRAMES**

**SERIES** : One Cloumn

**DATA FRAMES** : Entire Table

```python
import pandas as pd
ages = pd.Series([25,30,45,38])
print(ages)
```

```python
# We can Create Custom Tables
ages = pd.Series([25,30,45],
                 index=["Customer_1","Customer_2","Customer_3"])
print(ages)
```

```python
# DATA FRAMES
data = {
    "name":["Alice","Bob","Charlie"],
    "age":[25,35,45],
    "city":["Bangalore","Delhi","Mumbai"],
    "churn":[0,1,0]
}

# Create a Data Frame
df = pd.DataFrame(data) # Think DATA FRAME as as FEATURES (input)
print(df)
# FOR ML : ROWS = Samples/Observations & CLOUMNS = FEATURES /VARIABLES
```
