**ADVANCED CONCEPTS ( Lambda Function, MAP/FILTER / REDUCE )**

````python
from functools import reduce
# LAMBDA FUNCTIONS
add = lambda a, b :a+b
print(add(4,5))

# MAP : Map transforms every item in a list individually.
nums = [1,2,3,4]
# 1. MAP: Square each number
# Expected output: [1, 4, 9, 16, 25]
squared = list(map(lambda x: x ** 2, nums))
print("Mapped:", squared)

# 2. FILTER:Filter removes items from a list based on a condition.
#  Keep only even numbers:  Expected output: [2, 4]
evens = list(filter(lambda x: x % 2 == 0, nums))
print("Filtered:", evens)

# 3. REDUCE: combines all items in a list into a single value.
# Multiply all numbers together (1 * 2 * 3 * 4 * 5) and  Expected output: 120
product = reduce(lambda x, y: x * y, nums)
print("Reduced:", product)
````

# **OOPS in PYTHON**

1. **Class : A blueprint or template for creating objects.**
2. **Object: An instance of a class containing real data.**

````python
class Car:
  pass

# Object Creation
car1 = Car()
car2 = Car()

print(type(car2))
````

**CONSTRUCTOR**

**Pillars Of OOPS**

**1. Inheritance**: Allows a new class (child) to adopt attributes and methods from an existing class (parent).

*Benefit*: Reuses code.

````python
class vehicle:    # Parent class
  def Move(self):
    return "Moving Forward"

class Bike(vehicle):   # Child class inherits Vehicle
  pass

obj1 = Bike()
print(obj1.Move())   # Output: Moving forward
````

2. **Polymorphism**: Allows different classes to have methods with the same name but different behaviors.

      Benefit: Provides flexibility.

````python
class Dog:
  def Speak(self):
    print("Woof!")

class Cat:
  def Speak(self):
    print("MEOW!")

# Same method name, different behavior
for animal in ([Dog(), Cat()]):
  print(animal.Speak())

````

3. **Encapsulation** : Restricts direct access to methods and variables to prevent accidental modification.

     **Implementation**: Prefix an attribute with a ***double underscore*** __ to make it private.

````python
class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute as ( _ _) is used

    def get_balance(self):  # Controlled access
        return self.__balance

account = BankAccount(1000)
# print(account.__balance)  # Error! Cannot access directly.
print(account.get_balance())  # Output: 1000
````

**4. Abstraction:** Hides complex implementation details and only shows the essential features to the user.

  **Implementation**: Uses Python's built-in ***abc*** (Abstract Base Classes) module.

````python
from abc import ABC, abstractmethod

class Appliance(ABC):  # Abstract Class
    @abstractmethod
    def turn_on(self):
        pass

class Fan(Appliance):
    def turn_on(self):
        return "Fan is spinning"

my_fan = Fan()
print(my_fan.turn_on())  # Output: Fan is spinning
````

# `What is __init__ or self` ?

`__init__ is the initializer or constructor . It automatically runs whenever you create a new object to set up its initial data.`

`self  is a placeholder representing the specific object you are creating or working with at that moment.`






**Understanding** `self`



`When you define a class, you write code for objects that do not exist yet. self is a temporary name used inside the class to say, "whichever object is running this code right now."`



```
1. To store data: It attaches variables directly to the object so they can be accessed anywhere.

2. To identify the object: Python automatically passes the object into the method as the first argument.
```



**Understanding** `__init__`


```
It is a special Python method called a constructor or initializer.
You do not call it manually; Python triggers it behind the scenes the exact moment an object is born.
```


```
1. To accept parameters: It lets you pass custom data when creating an object.

2. To set initial state: It ensures every object starts with the necessary properties.
```




````python
class User:

    # 1. __init__ sets up the starting data for every new user
    def __init__(self, username, role):

        # 2. self connects the passed-in data to this specific user instance
        self.username = username  # Instance variable
        self.role = role          # Instance variable

    # 3. self lets other methods access the object's unique data
    def display_profile(self):
        return f"User: {self.username} | Role: {self.role}"

# --- CREATING OBJECTS ---

# When we create 'user1', Python maps 'self' to 'user1'
user1 = User("Alice", "Admin")

# When we create 'user2', Python maps 'self' to 'user2'
user2 = User("Bob", "Guest")

# --- VERIFYING THE OUTPUT ---

# Each object holds its own isolated data via self
print(user1.display_profile())  # Output: User: Alice | Role: Admin
print(user2.display_profile())  # Output: User: Bob | Role: Guest
````
