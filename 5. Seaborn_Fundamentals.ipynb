# **Seaborn** For **ML**

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np

n_customers = 50
np.random.seed(42) # Optional: ensures identical data every run

data = {
    'CustomerID': [f"CUST-{i:04d}" for i in range(1, n_customers + 1)],
    'Age': np.random.randint(18, 65, size=n_customers),
    'Tenure_Months': np.random.randint(1, 60, size=n_customers),
    'Monthly_Charges': np.round(np.random.uniform(29.99, 119.99, size=n_customers), 2),
    'Contract': np.random.choice(['Month-to-month', 'One year', 'Two year'], size=n_customers),
    'Churn': np.random.choice([0, 1], size=n_customers, p=[0.7, 0.3])
}

df = pd.DataFrame(data)

sns.scatterplot(
    data=df,
    x="Tenure_Months",
    y="Monthly_Charges"  # 3. Fixed the typo (was "Monthly_Cahrges")
)
plt.title("Tenure vs Monthly Charges")
plt.show()
```

```python
sns.histplot(
    data=df,
    x="Age",
    kde= True  # KDE = Kernel Density Distribution > Histogram Uses Bins and KDE uses smooth Probability-density estimate
)
plt.show()
```

```python
sns.countplot(
    data=df,
    x="Churn"
)
plt.show()
```

```python
# Stripplot : for Individual Observations
sns.stripplot(
    data=df,
    x="Churn",
    y="Monthly_Charges"
)
plt.show()
```

```python
# Swarm Plot : It avoids Overlap
sns.swarmplot(
    data=df,
    x="Tenure_Months",
    y="Monthly_Charges",
    hue="Churn"
)
plt.show()
```
